```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from models.baselines.mobilenet_baseline import LightweightCNNBaseline
from models.baselines.transformer_baseline import TransformerBaseline
from models.backbone.modules.dataset import ELPVDataset
import torch

train_ds = ELPVDataset(split="train")
x, y = train_ds[0]
x = x.unsqueeze(0)  # [1, 1, 300, 300]

print("===== LIGHTWEIGHT CNN (width_mult=0.25) =====")
cnn = LightweightCNNBaseline(in_channels=1, num_classes=4, width_mult=0.25)
out_cnn = cnn(x)
print("Output shape:", out_cnn.shape)  # expect [1, 4]
print("Param count:", sum(p.numel() for p in cnn.parameters()))

print("\n===== TRANSFORMER (embed_dim=96, depth=2, heads=4) =====")
tfm = TransformerBaseline(in_channels=1, patch_size=15, num_classes=4,
                           embed_dim=96, depth=2, num_heads=4)
out_tfm = tfm(x)
print("Output shape:", out_tfm.shape)  # expect [1, 4]
print("Param count:", sum(p.numel() for p in tfm.parameters()))
```
