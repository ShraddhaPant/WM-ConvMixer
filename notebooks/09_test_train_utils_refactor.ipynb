```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent) if "notebooks" in str(Path.cwd()) else str(Path.cwd()))

from train_utils import set_seed, get_class_weights, train_model
from models.backbone.modules.dataset import ELPVDataset
from models.backbone.convmixer import ConvMixer
from torch.utils.data import DataLoader
import torch

set_seed()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

train_ds = ELPVDataset(split="train")
mean, std = train_ds.computed_stats
val_ds = ELPVDataset(split="val", normalize_stats=(mean, std))

train_loader = DataLoader(train_ds, batch_size=32, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=32)
class_weights = get_class_weights(train_ds.labels)

model = ConvMixer(in_channels=1, dim=128, depth=8, kernel_size=9, patch_size=10, num_classes=4)
set_seed() 
train_model(model, "ConvMixer-refactor-check", train_loader, val_loader, class_weights, DEVICE)
```
